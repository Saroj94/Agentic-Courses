## **MIDDLEWARE**
    Middleware acts as an intermediary layer that intercepts, inspects, and modifies data or messages moving between an AI agent and external systems (like LLMs, tools, databases, or user interfaces).

**Core Functions of Middleware in Agentic Frameworks**

 1. Prompt & Input Manipulation: Intercepts a user request to inject system instructions, inject context from a database (RAG), or strip out sensitive PII (Personally Identifiable Information) before sending it to the LLM.  
 2. LLM Guardrails & Safety: Inspects the raw text generated by an agent to block harmful content, check for hallucinations, or ensure the output aligns with safe execution policies.
 3. Tool & API Interception: Wraps around tools to automatically log API payloads, handle rate-limiting or retry logic when an external service fails, and cache frequent queries to save costs.
 4. State Management & Logging: Captures trace logs of the agent's reasoning steps for debugging tools (like LangSmith or Phoenix) and updates the conversational memory/state database.
 5. Output Parsing & Validation: Automatically catches malformed JSON or invalid tool calls from the LLM, fixing it on the fly or sending a structured error message back to the agent to try again.
