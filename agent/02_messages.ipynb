```python
from langchain.messages import (SystemMessage, 
                                HumanMessage, 
                                AIMessage, 
                                ToolMessage)

##system messages
sys_message = SystemMessage("You are the coding assistent.")

message = [
    sys_message,
    HumanMessage("Help me to write the code")
]
```

```python
## human message with metadata
human_message_1 = HumanMessage(
    content="Write the python code example for loop.",
    name = "Saroj", ## an identifier for different users
    id = "msg_26" ## an identifier for different messages
)

human_message_2 = HumanMessage("Ok great tell about SQL query.")
```

```python
## AI messages
ai_message = AIMessage("I'd be happy to assist you.")

messages = [
    sys_message,
    human_message_1,
    ai_message,
    human_message_2
]

messages
```

```python
## AI message
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather", ## name of the tool
        "args": {"location": "Delhi"}, 
        "id": "calls_123"
    }]
)

## execute tool and create results
## call the tool and fetch the response through weather API
weather_response = "Sunney, 27•F" ## right now for demo purpose just hardcoded the response instead of running entire pipeline

## tool message
tool_message = ToolMessage(
    content=weather_response,
    tool_call_id = "calls_123"
)

##continue the converstion
messages = [
    HumanMessage("What is the weather in New Delhi"),
    ai_message, ## Model's tool call
    tool_message, ## tool execution results
]

## to model with the messages to get response from model
# response = model.invoke(messages)
# response
```
