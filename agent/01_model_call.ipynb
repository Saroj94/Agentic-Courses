## **STREAM over INVOKE**

- **STREAM**: it works faster than invoke, it doesn't wait for creating entire content instead displays it along with generating.
- **INVOKE**: it will generate the content with the help of model and displays it.

```python
for chunk in model.stream("write 5 lines of poem."):
    print(chunk.text, end="|", flush = True)
```

## **BATCHING**
Batching is the collection independent request or inputs to the model to significantly reduce the cost, and improve the performance. 
Also processing can be done in parallel. 

```python
model_res = model.batch([
    "Hello, openai how are you today!",
    "write me 5 lines of funny jokes.",
    "Which animal sleeps more than 8 hours?"
])

for res in model_res:
    print(res.text)
```

```python
model_res = model.batch([
    "Hello, openai how are you today!",
    "write me 5 lines of funny jokes.",
    "Which animal sleeps more than 8 hours?"
],
config={
    "max_concurrency":5, #limits to parallel calls
})

for res in model_res:
    print(res.text)
```
